# 决策树 Decision Tree

## 决策树简介
决策树是一种常用的监督学习算法，适用于分类和回归任务。它通过递归地将数据划分为更小的子集，形成一个树状结构，每个节点表示一个特征的测试，每个分支表示测试结果，每个叶节点表示一个类别或数值预测。

例如，对于“是否是猫”的分类问题，可以构造出决策树如下图：
![](./assets/decision-tree.png)

基础概念：
1. 决策节点（Decision Node）：表示一个特征的测试，如“脸是否是圆的？”
2. 叶节点（Leaf Node）：表示最终的类别或数值预测，如“是猫”或“不是猫”。

## 决策树的构建
决策树的构建通常采用递归分裂的方法，主要的流程就是不断地选择最佳特征进行划分，直到满足停止条件。

例如，对于一个分类“是否是猫”的数据集，构建决策树的步骤如下：
1. 选择“耳朵形状”分类，将数据集分为“尖耳朵”和“圆耳朵”两部分。
2. 对于“尖耳朵”部分，继续选择“脸形状”进行划分，得到“圆脸”和“方脸”两部分。
3. 对于“圆耳朵”部分，继续选择“胡须是否存在”进行划分，得到“有胡须”和“无胡须”两部分。
4. 重复上述步骤，直到某次划分后所有样本属于同一类别，或者没有更多特征可用。

## 决策树的学习过程
1. 如何选择特征进行划分？
    需要选择最大纯度、最小杂质的特征进行划分。
2. 停止条件
    - 所有样本属于同一类别
    - 达到预设的树深度
    - 纯度提升达到了预设阈值
    - 某个节点的样本数小于预设阈值

## 测量纯度的方法
信息熵（Information Entropy）是衡量数据纯度的常用方法。信息熵越低，数据越纯。计算信息熵的公式如下：
$$
H = - \sum_{i=1}^{c} p_i \log_2(p_i)
$$
其中，$H$表示数据集D的信息熵，$c$表示类别的数量，$p_i$表示类别$i$在数据集D中的比例。规定$0 \log_2(0) = 0$。

对于二分类问题，假设类别A的比例为$p_1$，类别B的比例为$p_0 = 1-p_1$，则信息熵可以表示为：
$$
H(p_1) = - p_1 \log_2(p_1) - p_0 \log_2(p_0)
$$

## 选择分割：信息增益
信息增益（Information Gain）是选择最佳特征进行划分的标准。信息增益表示通过某个特征划分数据集后，信息熵的减少量。计算信息增益的公式如下：
$$
IG = H(root) - \sum_{v \in Values} \frac{|N_v|}{|N|} H(p_v)
$$
其中，$IG$表示信息增益，$H(root)$表示划分前的数据集信息熵，$Values$表示特征的所有可能取值，$N_v$表示划分后属于取值$v$的样本数，$N$表示总样本数，$H(p_v)$表示划分后子集的信息熵。

对于二分类问题，假设某特征有两个取值，左子节点的样本比例为$p_1^{left}$，右子节点的样本比例为$p_1^{right}$，左子节点的权重为$w_1^{left}$，右子节点的权重为$w_1^{right}$；根节点的比例为$p_1^{root}$，则信息增益可以表示为：
$$
IG = H(p_1^{root}) - (w_1^{left} H(p_1^{left}) + w_1^{right} H(p_1^{right}))
$$

含义：即根的信息熵减去各个子节点的信息熵加权平均值。信息增益越大，表示该特征划分后数据集纯度提升越多，因此选择信息增益最大的特征进行划分。

同时，也可以根据信息增益来决定是否停止划分：如果信息增益小于预设阈值，则停止划分，避免过拟合。

## 决策树的学习算法
决策树的学习算法主要包括以下步骤：
1. 根节点放入所有训练样本
2. 计算当前节点的、全部特征的信息增益，选择信息增益最大的特征
3. 划分数据集，创建子节点
4. 对每个子节点，重复步骤2-3，直到满足停止条件：
    - 所有样本属于同一类别
    - 达到预设的树深度
    - 纯度提升（信息增益）达到了预设阈值
    - 某个节点的样本数小于预设阈值
5. 将叶节点标记为类别标签
