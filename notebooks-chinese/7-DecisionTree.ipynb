# 决策树 Decision Tree

## 决策树简介
决策树是一种常用的监督学习算法，适用于分类和回归任务。它通过递归地将数据划分为更小的子集，形成一个树状结构，每个节点表示一个特征的测试，每个分支表示测试结果，每个叶节点表示一个类别或数值预测。

例如，对于“是否是猫”的分类问题，可以构造出决策树如下图：
![](./assets/decision-tree.png)

基础概念：
1. 决策节点（Decision Node）：表示一个特征的测试，如“脸是否是圆的？”
2. 叶节点（Leaf Node）：表示最终的类别或数值预测，如“是猫”或“不是猫”。

## 决策树的构建
决策树的构建通常采用递归分裂的方法，主要的流程就是不断地选择最佳特征进行划分，直到满足停止条件。

例如，对于一个分类“是否是猫”的数据集，构建决策树的步骤如下：
1. 选择“耳朵形状”分类，将数据集分为“尖耳朵”和“圆耳朵”两部分。
2. 对于“尖耳朵”部分，继续选择“脸形状”进行划分，得到“圆脸”和“方脸”两部分。
3. 对于“圆耳朵”部分，继续选择“胡须是否存在”进行划分，得到“有胡须”和“无胡须”两部分。
4. 重复上述步骤，直到某次划分后所有样本属于同一类别，或者没有更多特征可用。

## 决策树的学习过程
1. 如何选择特征进行划分？
    需要选择最大纯度、最小杂质的特征进行划分。
2. 停止条件
    - 所有样本属于同一类别
    - 达到预设的树深度
    - 纯度提升达到了预设阈值
    - 某个节点的样本数小于预设阈值

## 测量纯度的方法
信息熵（Information Entropy）是衡量数据纯度的常用方法。信息熵越低，数据越纯。计算信息熵的公式如下：
$$
H = - \sum_{i=1}^{c} p_i \log_2(p_i)
$$
其中，$H$表示数据集D的信息熵，$c$表示类别的数量，$p_i$表示类别$i$在数据集D中的比例。规定$0 \log_2(0) = 0$。

对于二分类问题，假设类别A的比例为$p_1$，类别B的比例为$p_0 = 1-p_1$，则信息熵可以表示为：
$$
H(p_1) = - p_1 \log_2(p_1) - p_0 \log_2(p_0)
$$

## 选择分割：信息增益
信息增益（Information Gain）是选择最佳特征进行划分的标准。信息增益表示通过某个特征划分数据集后，信息熵的减少量。计算信息增益的公式如下：
$$
IG = H(root) - \sum_{v \in Values} \frac{|N_v|}{|N|} H(p_v)
$$
其中，$IG$表示信息增益，$H(root)$表示划分前的数据集信息熵，$Values$表示特征的所有可能取值，$N_v$表示划分后属于取值$v$的样本数，$N$表示总样本数，$H(p_v)$表示划分后子集的信息熵。

对于二分类问题，假设某特征有两个取值，左子节点的样本比例为$p_1^{left}$，右子节点的样本比例为$p_1^{right}$，左子节点的权重为$w_1^{left}$，右子节点的权重为$w_1^{right}$；根节点的比例为$p_1^{root}$，则信息增益可以表示为：
$$
IG = H(p_1^{root}) - (w_1^{left} H(p_1^{left}) + w_1^{right} H(p_1^{right}))
$$

含义：即根的信息熵减去各个子节点的信息熵加权平均值。信息增益越大，表示该特征划分后数据集纯度提升越多，因此选择信息增益最大的特征进行划分。

同时，也可以根据信息增益来决定是否停止划分：如果信息增益小于预设阈值，则停止划分，避免过拟合。

## 决策树的学习算法
决策树的学习算法主要包括以下步骤：
1. 根节点放入所有训练样本
2. 计算当前节点的、全部特征的信息增益，选择信息增益最大的特征
3. 划分数据集，创建子节点
4. 对每个子节点，重复步骤2-3，直到满足停止条件：
    - 所有样本属于同一类别
    - 达到预设的树深度
    - 纯度提升（信息增益）达到了预设阈值
    - 某个节点的样本数小于预设阈值
5. 将叶节点标记为类别标签



## 独热编码的分类特征
如果一个分类特征有多个取值（如颜色：红色、蓝色、绿色），可以使用独热编码（One-Hot Encoding）将其转换为多个二元特征。例如，颜色特征可以转换为三个二元特征：是否为红色、是否为蓝色、是否为绿色。这样，决策树可以更好地处理多分类特征。

## 连续特征的处理
对于连续特征（如年龄、收入），可以通过选择一个阈值将其划分为两个部分。例如，对于年龄特征，可以选择一个阈值（如30岁），将数据集划分为“年龄小于等于30岁”和“年龄大于30岁”两部分。决策树可以通过尝试不同的阈值来找到最佳划分点，使得信息增益最大化。

## 回归树
回归树（Regression Tree）是决策树的一种变体，适用于回归任务。与分类树不同，回归树的叶节点表示**数值预测**，而不是类别标签。例如，根据房屋的面积、位置等特征，预测房屋的价格。

回归树的构建过程类似于分类树，但使用方差减少（Variance Reduction）作为划分标准。方差减少表示通过某个特征划分数据集后，目标变量的方差减少量。计算方差减少的公式如下：
$$
VR = Var(root) - \sum_{v \in Values} \frac{|N_v|}{|N|} Var(p_v)
$$
其中，$VR$表示方差减少，$Var(root)$表示划分前的数据集方差，$Values$表示特征的所有可能取值，$N_v$表示划分后属于取值$v$的样本数，$N$表示总样本数，$Var(p_v)$表示划分后子集的方差。

方差的计算公式如下：
$$
Var = \frac{1}{N-1} \sum_{i=1}^{N} (y_i - \bar{y})^2
$$
其中，$N$表示样本数，$y_i$表示第$i$个样本的目标变量值，$\bar{y}$表示目标变量的均值。

## 使用多个决策树
使用单一的决策树，可能个别样本的变换会导致决策树结构的巨大变化，影响模型的稳定性和泛化能力。为了解决这个问题，可以使用集成学习的方法，结合多个决策树来提高模型的性能。

## 带替换的随机采样
带替换的随机采样是一种常用的集成学习方法。它通过对训练数据进行多次有放回的随机采样，生成多个不同的子数据集，然后在每个子数据集上训练一个决策树模型。具体过程为：在原始数据集中，随机选择样本进行采样，每次采样后将样本放回原始数据集，以便在后续采样中可能再次被选择。这样，每个子数据集可能包含重复的样本，同时也可能遗漏一些样本。最终得到的数据集大小与原始数据集相同，但样本分布有所不同。

## 随机森林
随机森林（Random Forest）是一种集成学习方法，通过结合多个决策树来提高模型的性能和稳定性。随机森林的构建过程如下：
1. 给定大小为$m$的训练集
2. 对于 $b=1$ 到 $B$：
    - 使用带替换的随机采样，从训练集中采样出一个大小为$m$的子数据集
    - 在该子数据集上训练一个决策树模型 $T_b$，在每个节点划分时，随机选择 $k$ 个特征进行划分
3. 最终得到 $B$ 棵决策树，形成随机森林模型
4. 对于分类任务，随机森林通过投票的方式进行预测；对于回归任务，随机森林通过取平均值的方式进行预测。

为了更加随机化，随机森林在每个节点划分时，如果有 $n$ 个特征可以选择，不是考虑所有特征，而是随机选择 $k < n$ 个特征进行划分。这样可以增加模型的多样性，减少过拟合的风险。一般可取 $k = \sqrt{n}$。

## XGBoost
XGBoost（Extreme Gradient Boosting）在选择分割时，不是选择单一的决策树，而是通过梯度提升的方法，结合多个弱分类器（通常是浅层决策树）来构建一个强分类器。XGBoost通过迭代地训练多个决策树，每次训练时，关注前一轮模型预测错误的样本，从而逐步提高模型的性能。XGBoost还引入了正则化项，防止过拟合，并且通过并行计算和缓存优化，提高了训练速度。
